```python
#!/usr/bin/env python3
# -*- coding: utf-8 -*-
"""
📦 GERADOR DE DOCUMENTAÇÃO PDF — FinSight AI v5.0
Para rodar no Google Colab: Copie este código em células e execute sequencialmente
Equipe: G9-BR-TEAM-20 | Hackathon OCI 2026

Gera 6 PDFs:
  01_PRD_Regras_Negocio.pdf
  02_API_REST_Documentacao.pdf
  03_Arquitetura_OCI.pdf
  04_README_Deploy.pdf
  05_Guia_Usuario.pdf
  06_Ferramentas_Instalacao_VM.pdf
"""

# ============================================================
# CÉLULA 1: Instalação de dependências
# ============================================================
# Cole no Colab e execute:
!pip install reportlab -q

# ============================================================
# CÉLULA 2: Upload dos seus arquivos (main.py e index.html)
# ============================================================
# No menu lateral esquerdo do Colab, clique no ícone 📁 (Arquivos)
# Clique em "Fazer upload para o armazenamento da sessão"
# Selecione: main.py, index.html, FinanceAI.pdf
# OU use o código abaixo para upload automático:

from google.colab import files
uploaded = files.upload()  # Vai abrir um dialog para você selecionar os arquivos

# ============================================================
# CÉLULA 3: Código completo de geração dos 6 PDFs
# ============================================================

from reportlab.lib import colors
from reportlab.lib.pagesizes import A4
from reportlab.platypus import SimpleDocTemplate, Paragraph, Spacer, Table, TableStyle, PageBreak
from reportlab.lib.styles import getSampleStyleSheet, ParagraphStyle
from reportlab.lib.units import cm
from reportlab.lib.enums import TA_LEFT, TA_CENTER, TA_JUSTIFY
import os

output_dir = "/content/documentacao_finsight"
os.makedirs(output_dir, exist_ok=True)

def create_pdf(filename, title, content_blocks):
    filepath = os.path.join(output_dir, filename)
    doc = SimpleDocTemplate(filepath, pagesize=A4,
                          rightMargin=2*cm, leftMargin=2*cm,
                          topMargin=2*cm, bottomMargin=2*cm)
    styles = getSampleStyleSheet()
    title_style = ParagraphStyle('CustomTitle', parent=styles['Heading1'], fontSize=24,
        textColor=colors.HexColor('#4f46e5'), spaceAfter=30, alignment=TA_CENTER, fontName='Helvetica-Bold')
    h2_style = ParagraphStyle('CustomH2', parent=styles['Heading2'], fontSize=14,
        textColor=colors.HexColor('#3730a3'), spaceAfter=12, spaceBefore=16, fontName='Helvetica-Bold')
    h3_style = ParagraphStyle('CustomH3', parent=styles['Heading3'], fontSize=12,
        textColor=colors.HexColor('#6366f1'), spaceAfter=8, spaceBefore=12, fontName='Helvetica-Bold')
    body_style = ParagraphStyle('CustomBody', parent=styles['BodyText'], fontSize=10,
        leading=14, alignment=TA_JUSTIFY, spaceAfter=8)
    code_style = ParagraphStyle('CodeStyle', parent=styles['Code'], fontSize=9, leading=12,
        leftIndent=10, textColor=colors.HexColor('#1e293b'), backColor=colors.HexColor('#f1f5f9'), borderPadding=8)
    note_style = ParagraphStyle('NoteStyle', parent=styles['BodyText'], fontSize=9, leading=13,
        leftIndent=10, textColor=colors.HexColor('#92400e'), backColor=colors.HexColor('#fffbeb'),
        borderPadding=8, borderColor=colors.HexColor('#f59e0b'), borderWidth=1)

    story = []
    story.append(Paragraph(title, title_style))
    story.append(Paragraph("<i>FinSight AI v5.0 — G9-BR-TEAM-20 | Hackathon OCI 2026</i>",
        ParagraphStyle('subtitle', parent=styles['Normal'], alignment=TA_CENTER,
                      textColor=colors.grey, fontSize=10, spaceAfter=20)))
    story.append(Spacer(1, 0.5*cm))

    for block in content_blocks:
        btype = block.get('type', 'text')
        text = block.get('text', '')
        if btype == 'h2':
            story.append(Paragraph(text, h2_style))
        elif btype == 'h3':
            story.append(Paragraph(text, h3_style))
        elif btype == 'code':
            story.append(Paragraph(f"<pre>{text}</pre>", code_style))
            story.append(Spacer(1, 0.3*cm))
        elif btype == 'note':
            story.append(Paragraph(f"<b>Nota:</b> {text}", note_style))
            story.append(Spacer(1, 0.3*cm))
        elif btype == 'table':
            data = block.get('data', [[]])
            t = Table(data, colWidths=block.get('widths', None))
            t.setStyle(TableStyle([
                ('BACKGROUND', (0,0), (-1,0), colors.HexColor('#4f46e5')),
                ('TEXTCOLOR', (0,0), (-1,0), colors.whitesmoke),
                ('ALIGN', (0,0), (-1,-1), 'LEFT'),
                ('FONTNAME', (0,0), (-1,0), 'Helvetica-Bold'),
                ('FONTSIZE', (0,0), (-1,0), 10),
                ('BOTTOMPADDING', (0,0), (-1,0), 12),
                ('BACKGROUND', (0,1), (-1,-1), colors.HexColor('#f8fafc')),
                ('GRID', (0,0), (-1,-1), 0.5, colors.HexColor('#e2e8f0')),
                ('FONTNAME', (0,1), (-1,-1), 'Helvetica'),
                ('FONTSIZE', (0,1), (-1,-1), 9),
                ('LEFTPADDING', (0,0), (-1,-1), 8),
                ('RIGHTPADDING', (0,0), (-1,-1), 8),
                ('BOTTOMPADDING', (0,1), (-1,-1), 8),
            ]))
            story.append(t)
            story.append(Spacer(1, 0.5*cm))
        elif btype == 'pagebreak':
            story.append(PageBreak())
        else:
            story.append(Paragraph(text, body_style))

    doc.build(story)
    return filepath

# ============================================================
# CONTEÚDO DOS 6 PDFs
# ============================================================

# ---------- PDF 1: PRD + Regras de Negócio ----------
blocks_prd = [
    {"type": "h2", "text": "1. Visão Geral do Produto"},
    {"type": "text", "text": "O <b>FinSight AI</b> é um assistente inteligente de saúde financeira desenvolvido para o Hackathon OCI 2026 pela equipe <b>G9-BR-TEAM-20</b>. A solução transforma dados financeiros brutos em insights acionáveis, classificando despesas automaticamente, perfilando o comportamento financeiro e gerando recomendações personalizadas."},
    {"type": "h2", "text": "2. Objetivos de Negócio"},
    {"type": "text", "text": "• <b>Automatizar</b> classificação de transações em 7 categorias (Alimentação, Transporte, Saúde, Moradia, Educação, Lazer, Serviços)."},
    {"type": "text", "text": "• <b>Perfilizar</b> usuário em: Saudável, Em Observação ou Em Risco."},
    {"type": "text", "text": "• <b>Recomendar</b> ações concretas contextualizadas ao perfil."},
    {"type": "text", "text": "• <b>Evoluir</b> acompanhamento via histórico persistente e gráficos."},
    {"type": "text", "text": "• <b>Simular</b> cenários hipotéticos sem impactar histórico real."},
    {"type": "h2", "text": "3. Regras de Negócio (RN)"},
    {"type": "h3", "text": "RN-001: Classificação de Transações"},
    {"type": "text", "text": "Toda transação é classificada em 7 categorias via ML (TF-IDF + classificador) com fallback heurístico por keywords."},
    {"type": "note", "text": "Fallback: valor > R$ 1.000 → Moradia (60%); R$ 200-1000 → Serviços (55%); < R$ 200 → Lazer (50%)."},
    {"type": "h3", "text": "RN-002: Score de Saúde Financeira (0-1000)"},
    {"type": "text", "text": "• <b>Endividamento (0-300 pts):</b> max(0, 300 - endiv × 3)"},
    {"type": "text", "text": "• <b>Poupança (50-250 pts):</b> Alta=250, Média=150, Baixa=50"},
    {"type": "text", "text": "• <b>Ratio Gasto/Renda (10-300 pts):</b> ≤30%=300; ≤50%=220; ≤70%=140; ≤90%=70; >90%=10"},
    {"type": "text", "text": "• <b>Bônus Perfil (0-150 pts):</b> Saudável=150, Observação=70, Risco=0"},
    {"type": "note", "text": "Score final limitado a 1000. Quanto maior, melhor a saúde financeira."},
    {"type": "h3", "text": "RN-003: Definição de Perfil Financeiro"},
    {"type": "table", "data": [
        ["Perfil", "Condições Heurísticas", "Score"],
        ["Saudável", "Endiv ≤30% + Poupança Alta/Média + Gastos ≤50%", "≥ 5"],
        ["Em Observação", "Endiv 30-50% ou Poupança Baixa ou Gastos 50-70%", "2-4"],
        ["Em Risco", "Endiv >50% ou Gastos >70% ou Poupança Baixa + Gastos >90%", "≤ 1"]
    ], "widths": [4*cm, 8*cm, 3*cm]},
    {"type": "h3", "text": "RN-004: Alertas de Metas"},
    {"type": "text", "text": "• Gasto ultrapassa meta → alerta crítico (vermelho)"},
    {"type": "text", "text": "• Gasto ≥90% da meta → alerta proximidade (amarelo)"},
    {"type": "text", "text": "• Endividamento >30% → alerta endividamento"},
    {"type": "text", "text": "• Gastos >90% renda → alerta risco; >70% → alerta atenção"},
    {"type": "h3", "text": "RN-005: Recomendações"},
    {"type": "text", "text": "Máximo 8 recomendações por análise. Ativadas por thresholds percentuais da renda por categoria (Lazer >15%, Moradia >35%, etc.)."},
    {"type": "h3", "text": "RN-006: CSV em Lote"},
    {"type": "text", "text": "Colunas: renda_mensal, nivel_endividamento, frequencia_poupanca, descricao, valor. Header opcional. Transações com mesma renda/endiv/poup são agrupadas."},
    {"type": "h2", "text": "4. KPIs"},
    {"type": "table", "data": [
        ["KPI", "Meta", "Medição"],
        ["Precisão Classificação", "> 85%", "Acurácia validação"],
        ["Tempo Resposta API", "< 2s", "Latência p95"],
        ["Taxa Fallback", "< 15%", "% heurística vs ML"],
        ["Disponibilidade", "> 99%", "Uptime OCI Compute"]
    ], "widths": [5*cm, 3*cm, 6*cm]},
]

pdf1 = create_pdf("01_PRD_Regras_Negocio.pdf", "PRD & Regras de Negócio<br/>FinSight AI v5.0", blocks_prd)
print(f"✅ PDF 1: {pdf1}")

# ---------- PDF 2: API REST ----------
blocks_api = [
    {"type": "h2", "text": "1. Visão Geral"},
    {"type": "text", "text": "API FastAPI com endpoints RESTful. Retorna JSON UTF-8. CORS liberado para desenvolvimento."},
    {"type": "h2", "text": "2. Schemas Pydantic"},
    {"type": "h3", "text": "Transacao"},
    {"type": "code", "text": "descricao: str (1-200 chars)\nvalor: float (> 0)"},
    {"type": "h3", "text": "AnaliseRequest"},
    {"type": "code", "text": "renda_mensal: float (>0)\nnivel_endividamento: float (0-100)\nfrequencia_poupanca: str (Alta|Media|Baixa)\ntransacoes: List[Transacao] (min 1)\nmetas: Dict[str, float] (opcional)"},
    {"type": "h2", "text": "3. Endpoints"},
    {"type": "h3", "text": "GET / — Root"},
    {"type": "text", "text": "Retorna metadados da API, versão e lista de endpoints."},
    {"type": "h3", "text": "GET /health"},
    {"type": "text", "text": "Health check: status, timestamp, models_loaded, models_count, model_dir, db_path, oci_bucket."},
    {"type": "h3", "text": "POST /analise-financeira"},
    {"type": "text", "text": "<b>Payload:</b> AnaliseRequest JSON. <b>Resposta:</b> AnaliseResponse completo com perfil, score, resumo, recomendações, alertas, explicabilidade. <b>Persistência:</b> Salva no SQLite."},
    {"type": "code", "text": "Fluxo: Validação → Classificação ML → Agregação → Métricas → Perfil → Alertas → Recomendações → Persistência → JSON"},
    {"type": "table", "data": [["Código", "Significado"],["200", "Sucesso"],["422", "Dados inválidos"],["500", "Erro interno"]], "widths": [3*cm, 4*cm]},
    {"type": "h3", "text": "POST /simular"},
    {"type": "text", "text": "Mesmo schema de /analise-financeira, mas <b>SEM persistência</b> no histórico. Ideal para cenários hipotéticos."},
    {"type": "h3", "text": "GET /classificar"},
    {"type": "text", "text": "Query params: descricao, valor. Retorna categoria e confiança de uma única transação."},
    {"type": "code", "text": "GET /classificar?descricao=Supermercado&valor=420.5\n→ {categoria: alimentacao, confianca: 0.94}"},
    {"type": "h3", "text": "POST /analise-batch-csv"},
    {"type": "text", "text": "Upload multipart de arquivo .csv. Colunas: renda_mensal, nivel_endividamento, frequencia_poupanca, descricao, valor. Agrupa por renda/endiv/poup."},
    {"type": "h3", "text": "GET /historico?limit=50"},
    {"type": "text", "text": "Lista análises persistidas (timestamp DESC). Limit 1-200. Retorna total + itens com todos os campos desserializados."},
    {"type": "h3", "text": "GET /relatorio/{item_id}"},
    {"type": "text", "text": "Recupera análise específica por ID. 404 se não existir."},
    {"type": "h3", "text": "DELETE /historico/{item_id}"},
    {"type": "text", "text": "Remove análise específica. Retorna {deleted: N}."},
    {"type": "h3", "text": "POST /historico/limpar"},
    {"type": "text", "text": "Remove TODAS as análises + VACUUM."},
    {"type": "h3", "text": "POST /reload-models"},
    {"type": "text", "text": "Hot-reload dos modelos ML do OCI Object Storage. Executa sync + gc.collect()."},
    {"type": "h2", "text": "4. Serviços Internos"},
    {"type": "text", "text": "<b>CategoriaService:</b> ML (TF-IDF + Classificador + LabelEncoder) → fallback keywords.<br/><b>PerfilService:</b> ML com mapeamento dinâmico de features → fallback heurístico. Calcula score 0-1000 e explicabilidade.<br/><b>AlertaService:</b> Alertas por metas, endividamento e ratio gasto/renda.<br/><b>RecomendacaoService:</b> Até 8 recomendações contextualizadas."},
    {"type": "h2", "text": "5. Tratamento de Erros"},
    {"type": "table", "data": [
        ["Cenário", "Comportamento"],
        ["ML indisponível", "Fallback heurístico"],
        ["OCI offline", "Usa cache local"],
        ["CSV mal formatado", "HTTP 400"],
        ["Renda ≤ 0", "HTTP 422"],
        ["Zero transações", "HTTP 422"]
    ], "widths": [6*cm, 8*cm]},
]

pdf2 = create_pdf("02_API_REST_Documentacao.pdf", "API REST — Documentação<br/>FastAPI | FinSight AI v5.0", blocks_api)
print(f"✅ PDF 2: {pdf2}")

# ---------- PDF 3: Arquitetura OCI ----------
blocks_arch = [
    {"type": "h2", "text": "1. Diagrama de Arquitetura"},
    {"type": "code", "text": "Cliente Web (index.html) ←HTTP→ FastAPI (OCI Compute/VM:8000)\n                                    ↓\n              ┌─────────────┬─────────────┬─────────────┐\n              ↓             ↓             ↓\n        OCI Object    SQLite Local    Modelos .pkl\n        Storage       (finsight.db)   (TF-IDF, RF)\n        (Bucket)                      Fallback: keywords"},
    {"type": "h2", "text": "2. Serviços OCI"},
    {"type": "h3", "text": "OCI Object Storage"},
    {"type": "text", "text": "<b>Bucket:</b> hackathon-one-g9-team-20<br/><b>Artefatos:</b> vetorizador_tfidf.pkl, modelo_categoria_producao.pkl, codificador_categorias.pkl, modelo_perfil_producao.pkl, codificador_perfil.pkl<br/><b>Sync:</b> Instance Principals auth, cache inteligente (compara tamanho), 3 retries."},
    {"type": "h3", "text": "OCI Compute"},
    {"type": "text", "text": "<b>Shape:</b> VM.Standard.E4.Flex (2 OCPUs, 4GB RAM)<br/><b>SO:</b> Ubuntu 22.04 / Oracle Linux 8<br/><b>Porta:</b> 8000 (Uvicorn)<br/><b>Service:</b> systemd finsight.service"},
    {"type": "h3", "text": "OCI IAM"},
    {"type": "text", "text": "Instance Principals Security Token Signer. Dynamic Group + Policy para leitura do bucket."},
    {"type": "h2", "text": "3. SQLite Schema"},
    {"type": "code", "text": "CREATE TABLE historico (\n  id INTEGER PRIMARY KEY AUTOINCREMENT,\n  timestamp TEXT, renda_mensal REAL,\n  nivel_endividamento REAL, frequencia_poupanca TEXT,\n  perfil_financeiro TEXT, probabilidade REAL,\n  score_saude INTEGER, total_gasto REAL,\n  percentual_renda_gasta REAL, comprometimento_gastos REAL,\n  resumo_gastos TEXT, transacoes TEXT,\n  recomendacoes TEXT, alertas TEXT,\n  explicabilidade TEXT, metas TEXT\n);"},
    {"type": "text", "text": "Campos JSON são serializados/desserializados via json.dumps/json.loads."},
    {"type": "h2", "text": "4. Diretórios"},
    {"type": "table", "data": [
        ["Diretório", "Caminho", "Propósito"],
        ["Base", "~/finsight-ai", "Raiz da aplicação"],
        ["Modelos", "~/finsight-ai/models/", "Cache .pkl do OCI"],
        ["Dados", "~/finsight-ai/data/", "SQLite + datasets"],
        ["DB", "~/finsight-ai/data/finsight.db", "Histórico"],
        ["Backend", "~/finsight-ai/backend/main.py", "FastAPI"],
        ["Frontend", "~/finsight-ai/index.html", "SPA"]
    ], "widths": [4*cm, 6*cm, 5*cm]},
    {"type": "h2", "text": "5. Variáveis de Ambiente"},
    {"type": "table", "data": [
        ["Variável", "Default"],
        ["OCI_BUCKET_NAME", "hackathon-one-g9-team-20"],
        ["OCI_BUCKET_NAMESPACE", "auto-detect"]
    ], "widths": [6*cm, 8*cm]},
    {"type": "h2", "text": "6. Segurança"},
    {"type": "text", "text": "• CORS: restrinja em produção<br/>• Validação Pydantic em todos os inputs<br/>• Retry OCI: 3 tentativas<br/>• GC pós-carregamento de modelos<br/>• Cache inteligente de modelos"},
]

pdf3 = create_pdf("03_Arquitetura_OCI.pdf", "Arquitetura & Infraestrutura<br/>Oracle Cloud Infrastructure", blocks_arch)
print(f"✅ PDF 3: {pdf3}")

# ---------- PDF 4: README + Deploy ----------
blocks_readme = [
    {"type": "h2", "text": "1. Instalação Local"},
    {"type": "code", "text": "git clone <repo>\ncd finsight-ai\npython3 -m venv venv\nsource venv/bin/activate\npip install fastapi uvicorn pydantic pandas numpy scikit-learn joblib"},
    {"type": "h2", "text": "2. Estrutura"},
    {"type": "code", "text": "finsight-ai/\n├── main.py          # API FastAPI\n├── index.html       # Frontend SPA\n├── models/          # .pkl (auto-criado)\n└── data/            # finsight.db (auto-criado)"},
    {"type": "h2", "text": "3. Execução"},
    {"type": "code", "text": "uvicorn main:app --host 0.0.0.0 --port 8000 --reload\n# Swagger: http://localhost:8000/docs"},
    {"type": "h2", "text": "4. Deploy OCI"},
    {"type": "h3", "text": "Passo 1: Compute Instance"},
    {"type": "text", "text": "Crie VM.Standard.E4.Flex (2 OCPUs, 4GB). Ubuntu 22.04. Abra porta 8000 no Security List."},
    {"type": "h3", "text": "Passo 2: Instance Principals"},
    {"type": "code", "text": "Dynamic Group: instance.compartment.id = '<ocid>'\nPolicy: allow dynamic-group finsight-dg to read objects\n  in compartment id where target.bucket.name='hackathon-one-g9-team-20'"},
    {"type": "h3", "text": "Passo 3: Upload Modelos"},
    {"type": "code", "text": "oci os object put --bucket-name hackathon-one-g9-team-20 \\\n  --file modelo.pkl --name models/modelo.pkl"},
    {"type": "h3", "text": "Passo 4: Deploy"},
    {"type": "code", "text": "ssh opc@<ip>\nsudo dnf install python3 python3-pip -y\npip3 install fastapi uvicorn pydantic pandas numpy scikit-learn joblib oci\n# Copie main.py e index.html\nnohup uvicorn main:app --host 0.0.0.0 --port 8000 &"},
    {"type": "h3", "text": "Passo 5: Systemd"},
    {"type": "code", "text": "[Unit]\nDescription=FinSight AI\nAfter=network.target\n\n[Service]\nUser=opc\nWorkingDirectory=/home/opc/finsight-ai\nExecStart=/usr/local/bin/uvicorn main:app --host 0.0.0.0 --port 8000\nRestart=always\n\n[Install]\nWantedBy=multi-user.target"},
    {"type": "h2", "text": "5. Exemplo cURL"},
    {"type": "code", "text": "curl -X POST http://localhost:8000/analise-financeira \\\n  -H 'Content-Type: application/json' \\\n  -d '{\"renda_mensal\":4500, \"nivel_endividamento\":25,\"frequencia_poupanca\":\"Media\", \"transacoes\":[{\"descricao\":\"Supermercado\",\"valor\":420}]}'"},
    {"type": "h2", "text": "6. Equipe G9-BR-TEAM-20"},
    {"type": "table", "data": [
        ["Nome", "Função"],
        ["Fabio C. Zinetti", "Data Scientist"],
        ["João P. R. Deodato", "Data Analyst"],
        ["Edson H. F. da Silva", "Data Analyst"],
        ["Luciano R. da Silva", "Back-End"],
        ["Rodrigo M. Veiga", "Back-End"],
        ["André Neves Xavier", "Back-End"]
    ], "widths": [5*cm, 8*cm]},
]

pdf4 = create_pdf("04_README_Deploy.pdf", "README & Guia de Deploy<br/>FinSight AI v5.0", blocks_readme)
print(f"✅ PDF 4: {pdf4}")

# ---------- PDF 5: Guia do Usuário ----------
blocks_user = [
    {"type": "h2", "text": "1. Guia Rápido"},
    {"type": "text", "text": "O FinSight AI tem 3 abas: <b>Análise</b>, <b>Simulador</b> e <b>Histórico</b>. Suporta PT/EN/ES e modo claro/escuro."},
    {"type": "h3", "text": "Aba Análise"},
    {"type": "text", "text": "Preencha renda, endividamento e poupança. Defina metas por categoria (opcional). Adicione transações. Clique 'Analisar'. Use 'Exemplo' para dados demo."},
    {"type": "h3", "text": "Aba Simulador"},
    {"type": "text", "text": "Teste cenários hipotéticos sem salvar no histórico. Útil para planejamento ('E se eu mudasse de apartamento?')."},
    {"type": "h3", "text": "Aba Histórico"},
    {"type": "text", "text": "Veja análises anteriores com gráfico de evolução do score. Exclua itens ou limpe tudo."},
    {"type": "h2", "text": "2. Entendendo Resultados"},
    {"type": "h3", "text": "Score (0-1000)"},
    {"type": "table", "data": [
        ["Faixa", "Significado"],
        ["800-1000", "Excelente — invista o excedente"],
        ["600-799", "Bom — pequenos ajustes"],
        ["400-599", "Regular — atenção necessária"],
        ["200-399", "Fraco — revisão urgente"],
        ["0-199", "Crítico — plano de recuperação"]
    ], "widths": [4*cm, 10*cm]},
    {"type": "h3", "text": "Perfil"},
    {"type": "table", "data": [
        ["Perfil", "Cor", "Significado"],
        ["Saudável", "Verde", "Hábitos excelentes"],
        ["Em Observação", "Amarelo", "Ajustes necessários"],
        ["Em Risco", "Vermelho", "Situação crítica"]
    ], "widths": [4*cm, 3*cm, 7*cm]},
    {"type": "h2", "text": "3. Confiança: 75,5% vs 94%"},
    {"type": "text", "text": "<b>⚠️ São duas confianças DIFERENTES:</b>"},
    {"type": "h3", "text": "Confiança 75,5% — Do PERFIL"},
    {"type": "text", "text": "É a probabilidade do modelo de que o perfil atribuído ('Em Observação') está correto. É agregada de TODA a análise (renda + endividamento + poupança + gastos)."},
    {"type": "text", "text": "• 75,5% = 75,5% de certeza de 'Em Observação', 24,5% divididos entre outros perfis.<br/>• Mais dados = maior confiança.<br/>• Fallback fixo: 88% (Saudável), 78% (Observação), 85% (Risco)."},
    {"type": "h3", "text": "Confiança 94% — Da CATEGORIA"},
    {"type": "text", "text": "É a probabilidade de UMA transação específica estar na categoria correta ('Supermercado' = 'Alimentação')."},
    {"type": "text", "text": "• 94% = alta certeza de classificação correta.<br/>• Calculada pelo TF-IDF + Classificador (probabilidade máxima entre 7 classes).<br/>• < 70% = descrição ambígua. Melhore a descrição."},
    {"type": "note", "text": "RESUMO: 75,5% é sobre SUA SAÚDE FINANCEIRA INTEIRA. 94% é sobre UMA TRANSAÇÃO ESPECÍFICA. São métricas independentes."},
    {"type": "h2", "text": "4. Dicas Avançadas"},
    {"type": "text", "text": "• Use o Simulador antes de decisões grandes.<br/>• Defina metas realistas (nem muito baixas nem altas).<br/>• Descrições detalhadas classificam melhor.<br/>• Acompanhe o histórico para ver evolução.<br/>• Exporte relatórios mensais via impressão."},
]

pdf5 = create_pdf("05_Guia_Usuario.pdf", "Guia do Usuário & Explicabilidade<br/>FinSight AI v5.0", blocks_user)
print(f"✅ PDF 5: {pdf5}")

# ---------- PDF 6: Ferramentas e Instalação na VM ----------
blocks_tools = [
    {"type": "h2", "text": "1. Stack Tecnológica Completa"},
    {"type": "text", "text": "O FinSight AI utiliza uma stack moderna e cloud-native, dividida em camadas de backend, machine learning, frontend e infraestrutura."},
    {"type": "h3", "text": "1.1 Backend — Python & FastAPI"},
    {"type": "table", "data": [
        ["Ferramenta", "Versão", "Função"],
        ["Python", "3.11+", "Linguagem principal do backend"],
        ["FastAPI", "latest", "Framework web assíncrono para API REST"],
        ["Uvicorn", "latest", "Servidor ASGI"],
        ["Pydantic", "latest", "Validação de dados e serialização JSON"],
        ["python-multipart", "latest", "Upload de arquivos (CSV em lote)"]
    ], "widths": [4*cm, 3*cm, 7*cm]},
    {"type": "h3", "text": "1.2 Machine Learning & Data Science"},
    {"type": "table", "data": [
        ["Ferramenta", "Versão", "Função"],
        ["scikit-learn", "latest", "TF-IDF, classificadores, LabelEncoder"],
        ["pandas", "latest", "Manipulação de dados tabulares"],
        ["numpy", "latest", "Computação numérica"],
        ["joblib", "latest", "Serialização de modelos .pkl"]
    ], "widths": [4*cm, 3*cm, 7*cm]},
    {"type": "h3", "text": "1.3 Cloud — OCI"},
    {"type": "table", "data": [
        ["Serviço", "Função"],
        ["OCI Compute (VM)", "Hospedagem da aplicação FastAPI + frontend"],
        ["OCI Object Storage", "Armazenamento centralizado dos modelos .pkl"],
        ["OCI IAM", "Autenticação via Instance Principals"]
    ], "widths": [5*cm, 9*cm]},
    {"type": "h3", "text": "1.4 Infraestrutura & Servidor Web"},
    {"type": "table", "data": [
        ["Ferramenta", "Função"],
        ["Nginx", "Reverse proxy, arquivos estáticos, SSL/TLS"],
        ["systemd", "Gerenciamento do serviço finsight (auto-restart)"],
        ["Certbot", "Certificados Let's Encrypt (HTTPS)"],
        ["SWAP", "Prevenção de OOM durante carregamento de modelos"]
    ], "widths": [4*cm, 10*cm]},
    {"type": "h3", "text": "1.5 Frontend — HTML/CSS/JS (Vanilla)"},
    {"type": "table", "data": [
        ["Tecnologia", "Origem", "Função"],
        ["Tailwind CSS", "CDN", "Framework CSS utilitário"],
        ["Chart.js", "CDN", "Gráficos interativos"],
        ["Lucide Icons", "CDN", "Ícones vetoriais"],
        ["Google Fonts (Inter)", "CDN", "Tipografia moderna"]
    ], "widths": [4*cm, 3*cm, 7*cm]},
    {"type": "h2", "text": "2. Instalação na VM OCI — Do Zero"},
    {"type": "h3", "text": "2.1 Pré-requisitos da VM"},
    {"type": "text", "text": "• <b>Shape:</b> VM.Standard.E4.Flex (mínimo 2 OCPUs, 4GB RAM)<br/>• <b>SO:</b> Ubuntu 22.04 LTS ou Oracle Linux 8<br/>• <b>Portas liberadas:</b> 22 (SSH), 80 (HTTP), 443 (HTTPS), 8000 (API direta)<br/>• <b>Armazenamento:</b> Mínimo 50GB boot volume<br/>• <b>Rede:</b> VCN com Internet Gateway e Security List configurada"},
    {"type": "h3", "text": "2.2 Criar a Instância no OCI Console"},
    {"type": "text", "text": "1. OCI Console → Compute → Instances → Create Instance<br/>2. Nome: finsight-ai-vm<br/>3. Shape: VM.Standard.E4.Flex (2 OCPUs, 4GB RAM)<br/>4. Image: Ubuntu 22.04 (Canonical)<br/>5. Networking: VCN + subnet pública<br/>6. Add SSH keys<br/>7. Boot volume: 50GB<br/>8. Create"},
    {"type": "h3", "text": "2.3 Configurar Security List (Firewall)"},
    {"type": "table", "data": [
        ["Tipo", "Porta", "Origem", "Descrição"],
        ["TCP", "22", "0.0.0.0/0", "SSH"],
        ["TCP", "80", "0.0.0.0/0", "HTTP — Nginx"],
        ["TCP", "443", "0.0.0.0/0", "HTTPS — Nginx + SSL"],
        ["TCP", "8000", "10.0.0.0/8", "API FastAPI (interno)"]
    ], "widths": [3*cm, 3*cm, 4*cm, 5*cm]},
    {"type": "h3", "text": "2.4 Configurar Instance Principals (IAM)"},
    {"type": "code", "text": "# 1. Crie uma Dynamic Group no OCI IAM:\n# Nome: finsight-dg\n# Matching Rule: instance.compartment.id = 'ocid1.compartment.oc1..xxxx'\n\n# 2. Crie uma Policy:\n# allow dynamic-group finsight-dg to read objects in compartment id\n#   where target.bucket.name='hackathon-one-g9-team-20'\n\n# 3. Anexe a Dynamic Group à sua instância Compute"},
    {"type": "note", "text": "Sem o Instance Principals configurado, o backend funcionará com fallback heurístico (sem modelos ML), reduzindo a precisão da classificação."},
    {"type": "h2", "text": "3. Script de Instalação Automatizada"},
    {"type": "h3", "text": "3.1 Estrutura de Diretórios na VM"},
    {"type": "code", "text": "/home/ubuntu/\n└── finsight-ai/\n    ├── backend/           # main.py + venv\n    ├── models/            # .pkl baixados do OCI\n    ├── data/              # SQLite (finsight.db)\n    └── venv/              # Virtualenv Python\n\n/var/www/\n└── finsight/            # index.html (nginx)"},
    {"type": "h3", "text": "3.2 Comandos de Instalação Manual"},
    {"type": "code", "text": "# SSH para a VM\nssh -i ~/.ssh/sua_chave.pem ubuntu@<IP_PUBLICO>\n\n# Atualizar sistema\nsudo apt update && sudo apt upgrade -y\n\n# Instalar Python, pip, nginx, git\nsudo apt install -y python3 python3-pip python3-venv nginx git curl\n\n# Criar diretórios\nmkdir -p ~/finsight-ai/backend ~/finsight-ai/models ~/finsight-ai/data\nsudo mkdir -p /var/www/finsight\n\n# Criar virtualenv\npython3 -m venv ~/finsight-ai/venv\nsource ~/finsight-ai/venv/bin/activate\n\n# Instalar dependências Python\npip install --upgrade pip\npip install fastapi uvicorn pydantic pandas numpy scikit-learn joblib oci python-multipart\n\n# Desativar venv\ndeactivate"},
    {"type": "h3", "text": "3.3 Copiar Arquivos para a VM"},
    {"type": "code", "text": "# Do seu computador local:\nscp -i ~/.ssh/sua_chave.pem main.py ubuntu@<IP>:~/finsight-ai/backend/\nscp -i ~/.ssh/sua_chave.pem index.html ubuntu@<IP>:/tmp/\n\n# Na VM:\nsudo cp /tmp/index.html /var/www/finsight/\nsudo chown -R www-data:www-data /var/www/finsight"},
    {"type": "h2", "text": "4. Configuração do Systemd"},
    {"type": "text", "text": "O systemd garante que a API reinicie automaticamente em caso de falha ou reboot da VM."},
    {"type": "code", "text": "sudo tee /etc/systemd/system/finsight.service > /dev/null << 'EOF'\n[Unit]\nDescription=FinSight AI API v5.0 - G9-BR-TEAM-20\nAfter=network.target\n\n[Service]\nUser=ubuntu\nWorkingDirectory=/home/ubuntu/finsight-ai/backend\nEnvironment=\"PATH=/home/ubuntu/finsight-ai/venv/bin\"\nEnvironment=\"API_PORT=8000\"\nEnvironment=\"API_HOST=0.0.0.0\"\nEnvironment=\"MODEL_DIR=/home/ubuntu/finsight-ai/models\"\nEnvironment=\"DATA_DIR=/home/ubuntu/finsight-ai/data\"\nEnvironment=\"OCI_BUCKET_NAME=hackathon-one-g9-team-20\"\nExecStart=/home/ubuntu/finsight-ai/venv/bin/uvicorn main:app --host 0.0.0.0 --port 8000\nRestart=always\nRestartSec=5\nStandardOutput=journal\nStandardError=journal\n\n[Install]\nWantedBy=multi-user.target\nEOF\n\nsudo systemctl daemon-reload\nsudo systemctl enable finsight\nsudo systemctl start finsight\nsudo systemctl status finsight"},
    {"type": "h2", "text": "5. Configuração do Nginx"},
    {"type": "h3", "text": "5.1 Modo HTTP (sem domínio)"},
    {"type": "code", "text": "sudo tee /etc/nginx/sites-available/finsight > /dev/null << 'EOF'\nserver {\n    listen 80;\n    server_name _;\n    location / { root /var/www/finsight; try_files $uri /index.html; }\n    location /docs { proxy_pass http://localhost:8000/docs; proxy_set_header Host $host; }\n    location /openapi.json { proxy_pass http://localhost:8000/openapi.json; proxy_set_header Host $host; }\n    location /health { proxy_pass http://localhost:8000/health; proxy_set_header Host $host; }\n    location /analise-financeira { proxy_pass http://localhost:8000/analise-financeira; proxy_set_header Host $host; }\n    location /simular { proxy_pass http://localhost:8000/simular; proxy_set_header Host $host; }\n    location /classificar { proxy_pass http://localhost:8000/classificar; proxy_set_header Host $host; }\n    location /historico { proxy_pass http://localhost:8000/historico; proxy_set_header Host $host; }\n    location /relatorio { proxy_pass http://localhost:8000/relatorio; proxy_set_header Host $host; }\n    location /reload-models { proxy_pass http://localhost:8000/reload-models; proxy_set_header Host $host; }\n}\nEOF\n\nsudo ln -sf /etc/nginx/sites-available/finsight /etc/nginx/sites-enabled/\nsudo rm -f /etc/nginx/sites-enabled/default\nsudo nginx -t\nsudo systemctl restart nginx"},
    {"type": "h3", "text": "5.2 Modo HTTPS (com domínio + Let's Encrypt)"},
    {"type": "code", "text": "sudo apt install -y certbot python3-certbot-nginx\n\n# Substitua example.com pelo seu domínio\nsudo tee /etc/nginx/sites-available/finsight > /dev/null << 'EOF'\nserver {\n    listen 80;\n    server_name example.com;\n    location / { return 301 https://$server_name$request_uri; }\n}\nserver {\n    listen 443 ssl;\n    server_name example.com;\n    ssl_certificate /etc/letsencrypt/live/example.com/fullchain.pem;\n    ssl_certificate_key /etc/letsencrypt/live/example.com/privkey.pem;\n    location / { root /var/www/finsight; try_files $uri /index.html; }\n    location /docs { proxy_pass http://localhost:8000/docs; proxy_set_header Host $host; }\n    location /analise-financeira { proxy_pass http://localhost:8000/analise-financeira; proxy_set_header Host $host; }\n    # ... demais locations iguais ao modo HTTP\n}\nEOF\n\nsudo certbot --nginx -d example.com --non-interactive --agree-tos -m admin@example.com"},
    {"type": "h2", "text": "6. Gerenciamento de SWAP (OOM Prevention)"},
    {"type": "text", "text": "Os modelos ML (.pkl) consomem memória RAM significativa ao serem carregados. O SWAP previne que o kernel mate o processo por falta de memória."},
    {"type": "code", "text": "# Verificar SWAP\nswapon --show\n\n# Criar arquivo de 2GB\nsudo fallocate -l 2G /swapfile || sudo dd if=/dev/zero of=/swapfile bs=1M count=2048\nsudo chmod 600 /swapfile\nsudo mkswap /swapfile\nsudo swapon /swapfile\necho '/swapfile none swap sw 0 0' | sudo tee -a /etc/fstab\n\n# Verificar\nfree -h"},
    {"type": "note", "text": "Recomenda-se 2GB de SWAP para shapes com 4GB RAM. Para 8GB+ RAM, 1GB é suficiente."},
    {"type": "h2", "text": "7. Script de Atualização Automática (ATUALIZAR_VM_v4.sh)"},
    {"type": "text", "text": "O script oficial de atualização executa 10 passos sequenciais:"},
    {"type": "table", "data": [
        ["Passo", "Ação", "Descrição"],
        ["0", "Gerenciar SWAP", "Cria/ativa SWAP de 2GB se não existir"],
        ["1", "Backup", "Cria backup de main.py e index.html"],
        ["2", "Dependências", "Atualiza pip e instala pacotes Python"],
        ["3", "Atualizar Backend", "Copia novo main.py para backend/"],
        ["4", "Atualizar Frontend", "Copia novo index.html para /var/www/finsight/"],
        ["5", "Verificar Modelos", "Checa se todos os .pkl existem localmente"],
        ["6", "Systemd", "Recria o arquivo de serviço finsight.service"],
        ["7", "Nginx", "Reconfigura sites-available (HTTP ou HTTPS)"],
        ["8", "Reiniciar", "Reload systemd, restart finsight + nginx"],
        ["9", "Verificação", "Health check na porta 8000"],
        ["10", "Status", "Exibe uso de memória (free -h)"]
    ], "widths": [2*cm, 4*cm, 8*cm]},
    {"type": "text", "text": "<b>Como usar:</b> Copie o script ATUALIZAR_VM_v4.sh para /home/ubuntu/ e execute <code>bash ATUALIZAR_VM_v4.sh</code>. O script é idempotente."},
    {"type": "h2", "text": "8. Backup e Rollback"},
    {"type": "code", "text": "# Backup manual\ncp ~/finsight-ai/backend/main.py ~/main.py.backup\ncp /var/www/finsight/index.html ~/index.html.backup\n\n# Rollback manual\ncp ~/main.py.backup ~/finsight-ai/backend/main.py\ncp ~/index.html.backup /var/www/finsight/index.html\nsudo systemctl restart finsight nginx"},
    {"type": "h2", "text": "9. Troubleshooting — Comandos Úteis"},
    {"type": "table", "data": [
        ["Problema", "Comando de Diagnóstico"],
        ["API não responde", "sudo journalctl -u finsight -f --no-pager"],
        ["Erro 502 Bad Gateway", "sudo nginx -t && sudo systemctl status nginx"],
        ["Modelos não carregam", "curl http://localhost:8000/health | jq ."],
        ["OOM / Memória cheia", "free -h && ps aux --sort=-%mem | head -10"],
        ["Porta 8000 ocupada", "sudo lsof -i :8000"],
        ["Logs da API", "sudo tail -f /var/log/syslog | grep finsight"],
        ["Testar endpoint", "curl -X POST http://localhost:8000/analise-financeira -H 'Content-Type: application/json' -d '{...}'"],
        ["Reiniciar tudo", "sudo systemctl restart finsight nginx"],
        ["Verificar SSL", "sudo certbot certificates"]
    ], "widths": [5*cm, 9*cm]},
    {"type": "h2", "text": "10. Variáveis de Ambiente"},
    {"type": "table", "data": [
        ["Variável", "Valor Padrão", "Onde Definir"],
        ["PATH", "~/finsight-ai/venv/bin", "systemd service"],
        ["API_PORT", "8000", "systemd service"],
        ["API_HOST", "0.0.0.0", "systemd service"],
        ["MODEL_DIR", "~/finsight-ai/models", "systemd service"],
        ["DATA_DIR", "~/finsight-ai/data", "systemd service"],
        ["OCI_BUCKET_NAME", "hackathon-one-g9-team-20", "systemd service"],
        ["OCI_BUCKET_NAMESPACE", "(auto-detect)", "Código Python"]
    ], "widths": [5*cm, 5*cm, 5*cm]},
    {"type": "h2", "text": "11. Checklist de Deploy"},
    {"type": "text", "text": "• VM criada com shape adequado (2 OCPUs, 4GB RAM)<br/>• Security List com portas 22, 80, 443 liberadas<br/>• Instance Principals configurado no IAM<br/>• Bucket OCI com modelos .pkl no prefixo models/<br/>• Python 3.11+ instalado<br/>• Virtualenv criado e dependências instaladas<br/>• main.py e index.html copiados para os diretórios corretos<br/>• Systemd service habilitado e iniciado<br/>• Nginx configurado e testado (nginx -t)<br/>• SWAP ativado (2GB)<br/>• Health check retorna HTTP 200<br/>• Frontend acessível via IP público"},
]

pdf6 = create_pdf("06_Ferramentas_Instalacao_VM.pdf", "Ferramentas, Dependências & Instalação na VM<br/>FinSight AI v5.0", blocks_tools)
print(f"✅ PDF 6: {pdf6}")

# ============================================================
# CÉLULA 4: Download dos PDFs
# ============================================================
from google.colab import files
import glob

print("\n" + "="*60)
print("📦 TODOS OS 6 PDFs GERADOS COM SUCESSO!")
print("="*60)

for pdf in sorted(glob.glob(output_dir + "/*.pdf")):
    size_kb = os.path.getsize(pdf) / 1024
    print(f"📄 {os.path.basename(pdf)} ({size_kb:.1f} KB)")

print("\n💡 Para baixar todos de uma vez, execute:")
print("   files.download('/content/documentacao_finsight/01_PRD_Regras_Negocio.pdf')")
print("   # ... repita para cada arquivo")
print("\n   OU use o painel lateral esquerdo (📁) para fazer download.")
```
